```python
import pandas as pd

# Варіант 8 — Полтава
wide = pd.DataFrame({
    "місто": ["Полтава"],
    "Січ": [77],
    "Лют": [73],
    "Бер": [66],
    "Кві": [56],
    "Тра": [47],
    "Чер": [41],
    "Лип": [37],
    "Сер": [39],
    "Вер": [49],
    "Жов": [63],
    "Лис": [75],
    "Гру": [81]
})

print(wide)
```

```python
tidy = wide.melt(
    id_vars="місто",
    var_name="місяць",
    value_name="хмарність"
)

print(tidy)
print("Форма:", tidy.shape)
```

```python
classification = {
    "місто": "номінальна",
    "місяць": "порядкова",
    "хмарність": "відносна"
}

for variable, scale in classification.items():
    print(f"{variable} — {scale} шкала")
```

```python
back = tidy.pivot(
    index="місто",
    columns="місяць",
    values="хмарність"
).reset_index()

back = back[wide.columns]

print(back)

print("Таблиці збігаються:", back.equals(wide))
```

```python
tidy.to_csv("variant.csv", index=False)

tidy.to_json(
    "variant.json",
    orient="records",
    force_ascii=False
)

reloaded_csv = pd.read_csv("variant.csv")
reloaded_json = pd.read_json("variant.json")

print("CSV збігається:", reloaded_csv.equals(tidy))
print("JSON збігається:", reloaded_json.equals(tidy))

print("\nВміст variant.csv:")
print(open("variant.csv", encoding="utf-8").read())

print("\nВміст variant.json:")
print(open("variant.json", encoding="utf-8").read())
```

```python
1. Чому «місто» — номінальна шкала, а «хмарність» — шкала відношень?

Стовпець «місто» є номінальною шкалою, оскільки назва міста використовується лише для позначення категорії та не має числового порядку. «Хмарність» є шкалою відношень, оскільки це числова величина з природним нульовим значенням, а співвідношення між значеннями має зміст.

2. Чому wide-таблиця з місяцями в стовпцях не є tidy?

Wide-таблиця не є tidy, тому що значення місяців використовуються як назви окремих стовпців. Після застосування melt() всі місяці переносяться в один стовпець «місяць», а відповідні значення хмарності — у стовпець «хмарність». Таким чином, кожен рядок відповідає одному спостереженню.

3. Чим CSV відрізняється від JSON?

CSV є простим табличним форматом, у якому дані зберігаються рядками та стовпцями, розділеними спеціальними символами. JSON має більш структурований формат і зберігає дані у вигляді об'єктів та масивів. CSV зручніший для таблиць, а JSON — для обміну структурованими даними між програмами.
```
