```python
import numpy as np

temps = np.array([-5, -4, 1, 10, 16, 19, 21, 20, 14, 8, 1, -3])

days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])

month_numbers = np.arange(1, 13)

print("temps:", temps)
print("shape:", temps.shape)
print("dtype:", temps.dtype)

print("\ndays_in_month:", days_in_month)
print("shape:", days_in_month.shape)
print("dtype:", days_in_month.dtype)

print("\nmonth_numbers:", month_numbers)
print("shape:", month_numbers.shape)
print("dtype:", month_numbers.dtype)
```

```python
temps_f = temps * 9 / 5 + 32

print("Температура у °F:")
print(temps_f)

warm_months = temps > 15

warm_days = days_in_month[warm_months].sum()

print("\nКількість днів у році, коли середньомісячна температура > 15°C:", warm_days)

print("\nСума днів у році:", days_in_month.sum())
print("Чи дорівнює 365:", days_in_month.sum() == 365)
```

```python
max_temp = temps.max()
max_index = temps.argmax()
max_month = month_numbers[max_index]

min_temp = temps.min()
min_index = temps.argmin()
min_month = month_numbers[min_index]

print("Найтепліший місяць:")
print("Номер місяця:", max_month)
print("Температура:", max_temp, "°C")
print("Індекс:", max_index)

print("\nНайхолодніший місяць:")
print("Номер місяця:", min_month)
print("Температура:", min_temp, "°C")
print("Індекс:", min_index)

below_zero_months = month_numbers[temps < 0]

print("\nМісяці з температурою нижче 0°C:")
print(below_zero_months)

sorted_temps = np.sort(temps)

print("\nТемператури за зростанням:")
print(sorted_temps)

print("\nОригінальний масив temps:")
print(temps)
```

```python
sun_hours = np.array([
    55, 65, 90, 140, 190, 230,
    245, 225, 165, 110, 65, 50
])

data = np.array([
    temps,
    days_in_month,
    sun_hours
])

print("Двовимірний масив:")
print(data)

print("\nФорма масиву:")
print(data.shape)

row_sums = data.sum(axis=1)

row_means = data.mean(axis=1)

print("\nСума по кожному рядку:")
print(row_sums)

print("\nСереднє по кожному рядку:")
print(row_means)

column_sums = data.sum(axis=0)

print("\nСума по кожному стовпцю:")
print(column_sums)
```

```python
rows = np.array([
    temps,
    days_in_month
])

print("rows:")
print(rows)

print("\nФорма rows:", rows.shape)

correction = np.array([1, 0])

print("\ncorrection:")
print(correction)

print("Форма correction:", correction.shape)

try:
    result = rows + correction
    print(result)
except ValueError as e:
    print("\nПомилка:")
    print(e)

correction_reshaped = correction.reshape(2, 1)

print("\nНова форма correction:")
print(correction_reshaped.shape)

result = rows + correction_reshaped

print("\nРезультат rows + correction.reshape(2, 1):")
print(result)
```
