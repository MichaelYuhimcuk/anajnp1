```python
import pandas as pd

months = [
    "Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
    "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"
]

avg_temp = [-5, -4, 1, 10, 16, 19, 21, 20, 14, 8, 1, -3]

precip = [38, 36, 34, 36, 46, 56, 60, 48, 36, 36, 44, 44]

df = pd.DataFrame({
    "month": months,
    "avg_temp": avg_temp,
    "precip": precip
})

df.head()
```

```python
print(
    df[
        (df["avg_temp"] < 0) |
        (df["precip"] > 80)
    ]
)
```

```python
df["avg_temp_f"] = df["avg_temp"] * 9 / 5 + 32

print(df)
```

```python
average_precip = df["precip"].mean()

df["is_wet"] = df["precip"] > average_precip

print(df)
```

```python
def get_season(month_number):
    if month_number in [12, 1, 2]:
        return "зима"
    elif month_number in [3, 4, 5]:
        return "весна"
    elif month_number in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"


df["season"] = df.index.to_series().apply(get_season)

print(df)
```
