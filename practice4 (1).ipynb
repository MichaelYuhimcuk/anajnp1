```python
import numpy as np
import pandas as pd

orders = pd.DataFrame([
    {"дата": "2026-09-01", "товар": "Навушники", "місто": "Полтава",        "ціна": 700,        "кількість": 4},
    {"дата": "2026-09-02", "товар": "Навушники", "місто": " Полтава ",      "ціна": "700 грн",   "кількість": 5},
    {"дата": "2026-09-03", "товар": "Навушники", "місто": "ПОЛТАВА",        "ціна": 720,        "кількість": np.nan},
    {"дата": "2026-09-04", "товар": "Навушники", "місто": "Полтава",        "ціна": "680 грн",   "кількість": 3},
    {"дата": "2026-09-05", "товар": "Навушники", "місто": "Полтава",        "ціна": 710,        "кількість": np.nan},
    {"дата": "2026-09-06", "товар": "Навушники", "місто": "ПОЛТАВА",        "ціна": 690,        "кількість": 4},
    {"дата": "2026-09-07", "товар": "Навушники", "місто": "Полтава",        "ціна": 5600,       "кількість": 4},
    {"дата": "2026-09-07", "товар": "Навушники", "місто": "Полтава",        "ціна": 5600,       "кількість": 4},
])

print("Брудний набір даних:")
display(orders)
```

```python
print("Кількість пропусків:")
print(orders.isna().sum())
```

```python
median_quantity = orders["кількість"].median()

print("Медіана кількості:", median_quantity)

orders["кількість"] = orders["кількість"].fillna(median_quantity)

print("Після заповнення пропусків:")
display(orders)
```

```python
print("Кількість повних дублікатів:")
print(orders.duplicated().sum())

print("\nДублікати:")
display(orders[orders.duplicated()])
```

```python
print("Кількість дублікатів за датою та товаром:")
print(orders.duplicated(subset=["дата", "товар"]).sum())

display(orders[orders.duplicated(subset=["дата", "товар"], keep=False)])
```

```python
orders = orders.drop_duplicates(subset=["дата", "товар"])

print("Кількість рядків після видалення дубліката:", len(orders))

display(orders)
```

```python
orders["ціна"] = (
    orders["ціна"]
    .astype(str)
    .str.replace(" грн", "", regex=False)
    .astype(float)
)

print("Тип даних ціни:")
print(orders["ціна"].dtype)

display(orders)
```

```python
orders["місто"] = orders["місто"].str.strip().str.lower()

print("Унікальні назви міст після очищення:")
print(orders["місто"].unique())
```

```python
orders["місто"] = orders["місто"].replace({
    "полтава": "Полтава"
})

print("Унікальні назви міст:")
print(orders["місто"].unique())

display(orders)
```

```python
q1 = orders["ціна"].quantile(0.25)
q3 = orders["ціна"].quantile(0.75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

print("Q1 =", q1)
print("Q3 =", q3)
print("IQR =", iqr)
print("Нижня межа =", lower)
print("Верхня межа =", upper)
```

```python
outliers = orders[
    (orders["ціна"] < lower) |
    (orders["ціна"] > upper)
]

print("Викиди:")
display(outliers)
```

```python
orders_clean = orders[
    (orders["ціна"] >= lower) &
    (orders["ціна"] <= upper)
].copy()

print("Очищений набір:")
display(orders_clean)
```

```python
1. Чому середнє може спотворити стандартне відхилення?

Заповнення пропуску середнім може зменшити або змінити стандартне відхилення, оскільки пропущене значення замінюється значенням, яке знаходиться саме в центрі набору. У результаті штучно зменшується розкид даних. Медіана є більш стійкою до викидів і нетипових значень, тому в цьому наборі вона є кращим варіантом.

2. Різниця між duplicated() без subset і з subset

duplicated() без subset шукає рядки, у яких повністю збігаються всі стовпці. Якщо вказати subset, перевіряються тільки вибрані стовпці. Наприклад, два замовлення можуть відрізнятися ціною або кількістю, але мати однакову дату та товар. Тоді при перевірці за subset=["дата", "товар"] вони будуть вважатися дубльованими.

3. Чому не можна автоматично видаляти всі значення поза IQR?

Значення за межами IQR не обов'язково є помилками. Це може бути реальне рідкісне спостереження. Наприклад, клієнт справді міг зробити дуже велике замовлення. Тому IQR потрібно використовувати для виявлення підозрілих значень, а остаточне рішення про видалення потрібно приймати після аналізу предметної області.
```
