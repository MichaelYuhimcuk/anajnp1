```python
import pandas as pd

print("\n" + "=" * 60)
print("ЗАВДАННЯ 1. describe()")
print("=" * 60)

poltava = pd.Series(
    [84, 81, 76, 69, 63, 61, 59, 61, 67, 74, 80, 85],
    name="Вологість"
)

print("Місто: Полтава")
print("Регіон: Центр")
print("\nДані за місяцями:")
print(poltava)

print("\nРезультат describe():")
print(poltava.describe())
```

```python
print("\n" + "=" * 60)
print("ЗАВДАННЯ 2. Ручне обчислення")
print("=" * 60)

rozmakh = poltava.max() - poltava.min()

var_ddof1 = poltava.var(ddof=1)
var_ddof0 = poltava.var(ddof=0)

std = poltava.std(ddof=1)

q1 = poltava.quantile(0.25)
q3 = poltava.quantile(0.75)

iqr = q3 - q1

cv = std / poltava.mean() * 100

print(f"Розмах: {rozmakh}")
print(f"Дисперсія ddof=1: {var_ddof1}")
print(f"Дисперсія ddof=0: {var_ddof0}")
print(f"Стандартне відхилення: {std}")
print(f"Q1 (25%): {q1}")
print(f"Q3 (75%): {q3}")
print(f"IQR: {iqr}")
print(f"Коефіцієнт варіації: {cv:.2f}%")

print("\nПеревірка квартилів:")
print("Q1 збігається:",
      q1 == poltava.describe()["25%"])
print("Q3 збігається:",
      q3 == poltava.describe()["75%"])
```

```python
print("\n" + "=" * 60)
print("ЗАВДАННЯ 3. Порівняння регіонів")
print("=" * 60)

months = [
    "Січ", "Лют", "Бер", "Кві",
    "Тра", "Чер", "Лип", "Сер",
    "Вер", "Жов", "Лис", "Гру"
]

data = {
    "Київ": (
        "Центр",
        [84, 82, 78, 70, 65, 63, 61, 63, 68, 75, 81, 85]
    ),

    "Львів": (
        "Захід",
        [85, 83, 80, 73, 68, 65, 63, 65, 70, 77, 82, 86]
    ),

    "Одеса": (
        "Південь",
        [82, 80, 78, 74, 71, 69, 67, 68, 72, 76, 80, 83]
    ),

    "Харків": (
        "Схід",
        [83, 80, 75, 68, 62, 60, 58, 60, 66, 73, 79, 84]
    ),

    "Дніпро": (
        "Схід",
        [82, 79, 74, 67, 61, 59, 57, 59, 65, 72, 78, 83]
    ),

    "Чернігів": (
        "Центр",
        [85, 83, 79, 71, 66, 63, 61, 63, 69, 76, 82, 86]
    ),

    "Івано-Франківськ": (
        "Захід",
        [86, 84, 81, 75, 70, 67, 65, 67, 72, 78, 83, 87]
    ),

    "Полтава": (
        "Центр",
        [84, 81, 76, 69, 63, 61, 59, 61, 67, 74, 80, 85]
    ),

    "Суми": (
        "Схід",
        [85, 82, 77, 70, 64, 62, 60, 62, 68, 75, 81, 86]
    ),

    "Ужгород": (
        "Захід",
        [84, 82, 79, 73, 68, 66, 64, 66, 71, 77, 82, 85]
    )
}

rows = []

for city, (region, values) in data.items():
    row = {
        "Місто": city,
        "Регіон": region
    }

    for month, value in zip(months, values):
        row[month] = value

    rows.append(row)

df = pd.DataFrame(rows)

df["Середня річна вологість"] = df[months].mean(axis=1)

print("\nПовний DataFrame:")
print(df)

center = df[
    df["Регіон"] == "Центр"
]["Середня річна вологість"]

print("\nСтатистика регіону Центр:")
print(center.describe())

east = df[
    df["Регіон"] == "Схід"
]["Середня річна вологість"]

print("\nСтатистика регіону Схід:")
print(east.describe())

cv_center = center.std() / center.mean() * 100
cv_east = east.std() / east.mean() * 100

print("\nКоефіцієнт варіації:")
print(f"Центр: {cv_center:.2f}%")
print(f"Схід: {cv_east:.2f}%")
```

```python
print("\n" + "=" * 60)
print("ЗАВДАННЯ 4. Форма розподілу")
print("=" * 60)

mean_poltava = poltava.mean()
median_poltava = poltava.median()

print(f"Середнє: {mean_poltava:.2f}")
print(f"Медіана: {median_poltava:.2f}")
print(
    f"Різниця між середнім і медіаною: "
    f"{abs(mean_poltava - median_poltava):.2f}"
)

print("\nВисновок:")
print(
    "Середнє та медіана дуже близькі, "
    "тому розподіл даних Полтави ближчий до симетричного."
)
print(
    "Середнє є репрезентативною мірою положення, "
    "оскільки сильного розходження між середнім і медіаною немає."
)
```

```python
print("\n" + "=" * 60)
print("ЗАВДАННЯ 5. Порівняння варіативності")
print("=" * 60)

lviv = pd.Series(
    [85, 83, 80, 73, 68, 65, 63, 65, 70, 77, 82, 86],
    name="Вологість"
)

cv_poltava = poltava.std() / poltava.mean() * 100
cv_lviv = lviv.std() / lviv.mean() * 100

print("Полтава:")
print(f"Середнє: {poltava.mean():.2f}")
print(f"Std: {poltava.std():.2f}")
print(f"CV: {cv_poltava:.2f}%")

print("\nЛьвів:")
print(f"Середнє: {lviv.mean():.2f}")
print(f"Std: {lviv.std():.2f}")
print(f"CV: {cv_lviv:.2f}%")
```

```python
1. У describe() рядок std відповідає стандартному відхиленню.
25% і 75% — це перший та третій квартилі.
min і max — мінімальне та максимальне значення.
Різниця між max і min утворює розмах.

2. Заповнення пропуску середнім штучно зменшує std,
тому що середнє значення не додає нового розкиду.
Наприклад, [10, 20, 30, 40] має std приблизно 12.91.
Після заміни одного значення середнім 25 отримуємо
[10, 20, 30, 25], де std приблизно 8.54.

3. Сирий std показує абсолютний розкид і залежить від
масштабу середнього значення. Коефіцієнт варіації
CV = std / mean * 100% показує відносний розкид,
тому він зручний для порівняння різних груп.

4. Для Полтави середнє 71.67%, а медіана 71.50%.
Вони дуже близькі, тому розподіл ближчий до симетричного.
Середнє є достатньо репрезентативною мірою положення.
```
